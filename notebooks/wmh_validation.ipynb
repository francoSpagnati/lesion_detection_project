# WMH validation

Verifica locale del dataset condiviso e configurazione nnU-Net. Nessun download viene eseguito.

```python
from pathlib import Path
import json
import nibabel as nib

PROJECT = Path('/home/ccantini/lesion_detection_project')
cfg = json.loads((PROJECT/'configs/datasets.json').read_text())
wmh = Path(cfg['datasets']['WMH']['path'])
files = sorted(wmh.rglob('*.nii.gz'))
print('WMH path:', wmh)
print('NIfTI files:', len(files))
```

```python
for p in files[:10]:
    img = nib.load(str(p), mmap=True)
    print(p.name, img.shape, tuple(round(float(x), 4) for x in img.header.get_zooms()[:3]))
```
